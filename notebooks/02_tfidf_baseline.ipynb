# 02 — TF-IDF Baseline (M1)

Runs `scripts/run_tfidf.py`: fits a scikit-learn `TfidfVectorizer` (sublinear TF, L2 norm, smoothed IDF — `configs/tfidf.yaml`) on the full NFCorpus corpus, ranks the top-100 documents for every TEST-split query, evaluates against the real test qrels with `pytrec_eval`, and writes `results/{runs,metrics,manifests}/*tfidf*`.

All retrieval/evaluation logic lives in `src/biomedical_ir/{tfidf,evaluation}.py` — this notebook only orchestrates.

```python
# If running on Colab, clone the repo and install deps. Skipped automatically
# when already inside a local checkout (REPO_ROOT / 'src' already importable).
import os, sys, subprocess

IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    if not os.path.exists('biomedical-hybrid-ir'):
        subprocess.run(['git', 'clone', 'https://github.com/Arungharami/biomedical-hybrid-ir'], check=True)
    os.chdir('biomedical-hybrid-ir')
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', '.'], check=True)

sys.path.insert(0, os.path.join(os.getcwd(), 'src'))
print('cwd:', os.getcwd())
```

```python
import time
start = time.perf_counter()
result = subprocess.run([sys.executable, 'scripts/run_tfidf.py'], cwd=os.getcwd())
print(f'\nexit code: {result.returncode}, elapsed: {time.perf_counter()-start:.1f}s')
assert result.returncode == 0, 'Script failed -- see output above.'
```
